## dtypes
Son los tipos de datos de pandas. Basicamente indican el tipo de operaciones que se pueden hacer con toda la serie

```python
# ejemplo df con columnas/series de tipos distintos
columnas1 = ['TipoEntero','TipoDecimal','TipoCadena','TipoBooleano','TipoMixto']
datos1 = [
    [1,1.0,'A',True,'a'],
    [2,1.5,'B',False,15],
    [3,-1.5,'C',False,True]
]
df1 = pd.DataFrame(columns=columnas1, data=datos1)
print(df1)
sep()
print(df1['TipoEntero'])
sep()
print(df1['TipoDecimal'])
sep()
print(df1['TipoCadena'])
sep()
print(df1['TipoBooleano'])
sep()
print(df1['TipoMixto'])
```

Podemos ver el tipo de datos (dtype) de cada columna de nuestro dataframe mediante su atributo dtypes

```python
df1.dtypes
```

Podemos hacer operaciones sobre SERIES que son de cierto tipo de dtype compatibles entre si. Las operaciones actuan elemento por elemento 

```python
# recordando
print('df:')
print(df)
sep()
print('df1:')
print(df1)
sep()

#Ejemplo de operaciones entre series
df['Edad'] + df1['TipoDecimal']
```

```python
# tambien puedo usar para operaciones basicas
df1["TipoEntero"] * 10 + 1000
```

Como seleccionar un valor de una SERIE

```python
print("df1['TipoEntero'][0]: ",df1['TipoEntero'][0])
print("df1['TipoEntero'][1]: ",df1['TipoEntero'][1])

# esto no funcionara, porque pandas usa su indice automatico (RangeIndex) que va de 0 a N (sin numeros negativos):
#   print(df1['TipoEntero'][-1]) ### DA ERROR
# una forma de obtener ese dato es usar 'iloc' (integer location) que nos permite obtener la posicion fisica asi como con las listas en py
print("df1['TipoEntero'].iloc[-1]: ",df1['TipoEntero'].iloc[-1])
```

Las modificaciones que se realizan sobre un dataframe no son permanentes al menos que sobre escribamos ese mismo dataframe

```python
df1.drop("TipoEntero",axis=1) 
```

```python
# Nota: axis=1 sirve para borrar toda la columna (Serie) . Si pongo axis=0 lo que hace es buscar el indice y borrar una fila
df1.drop(1,axis=0) 
```

```python
# volviendo al ejemplo anterior, aca podemos ver como mi df1 NO fue modificado. Esto es porque .drop me genero un NUEVO DATAFRAME con el cambio que le pedi
df1
```

si realmente queremos modificar nuestro df1...

```python
# primero lo voy a backupear 
df1_backup = df1.copy() # ojo, el backup no funciona sin el .copy
```

podemos hacerlo asi:

```python
df1 = df1.drop('TipoCadena',axis=1) 
df1
```

```python
# voy a restaurarlo
df1 = df1_backup
df1
```

o podriamos usar INPLACE:

```python
df1.drop('TipoCadena',axis=1,inplace=True) 
df1
```

```python
# voy a restaurarlo
df1 = df1_backup
df1
```
